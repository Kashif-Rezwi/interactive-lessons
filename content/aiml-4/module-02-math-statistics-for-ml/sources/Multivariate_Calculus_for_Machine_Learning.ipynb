#**PART 2 — Partial Derivatives**

**Definition**

- For multivariable functions:

$$f(x,y,z)$$

- a partial derivative measures:

  - Rate of change with respect to ONE variable while keeping others fixed.

**Example**

Suppose:

$$f(x,y)=x^
2
+3y$$

**Partial derivative w.r.t x:**

$$\frac{∂f}
{∂x}=2x$$

**Treat y as constant.**

- Partial Derivative w.r.t y

$$\frac{∂f}
{∂y}	​
=3$$

Treat x as constant.



**Intuition**

- A partial derivative measures:

  - Slope along one direction while freezing all others.

**Mountain Analogy**

- Imagine standing on a mountain.

**You can:**

- move east-west
- move north-south

Each direction has different slope.

- Partial derivative:

  - measures slope along chosen direction.

**Geometric Understanding**

- For function:

$$z=f(x,y)$$

**Partial derivatives describe:**

- local tilt of surface
- directional steepness


**Higher-Dimensional Perspective**

- In ML:

  - parameters may be millions-dimensional

- Partial derivatives generalize naturally:

$$\frac{∂L}{∂w_
i}$$
	​


**measures:**

- sensitivity of loss to parameter i


**ML Interpretation**

- Large partial derivative:

  - parameter strongly affects loss

**Small partial derivative:**

- parameter has weak local effect

**Python Example**

```python
def f(x, y):
    return x**2 + 3*y

x = 2
y = 4

df_dx = 2*x
df_dy = 3

print(df_dx, df_dy)
```

#**PART 3 — Gradient**

**Definition**

- The Gradient is a vector containing all partial derivatives.

For:

$$f(x,y)$$

gradient:

$$∇f=[\frac{
∂f}{∂x},
\frac{∂f}
{∂y}	​
]$$



**Gradient Intuition**

- Gradient tells:

  - how function changes
  - most sensitive direction
  - steepest increase direction


**Key Property**
- Gradient points in direction of steepest increase.

- This is one of the most important ideas in optimization.


**Geometric Meaning**

- Imagine standing on mountain surface.

**Gradient arrow points:**

- uphill fastest direction

**Negative gradient points:**

- downhill fastest direction


**Gradient Magnitude**

- Magnitude of gradient tells:

  - How steep the surface is.

**Large magnitude:**

- steep slope

**Small magnitude:**

- flat region


**Gradient Descent**
- Core optimization algorithm in ML:

$$θ:=θ−η∇f(θ)$$


**where:**

- θ = parameters
- η = learning rate
- ∇f = gradient


**Intuition**
- Gradient       → uphill
- Negative Gradient → downhill

**Optimization minimizes loss by moving:**

- opposite to gradient.

**Python Example — Gradient Descent**

```python
x = 5
lr = 0.1

for _ in range(20):
    grad = 2*x
    x = x - lr * grad

print(x)
```

#**PART 4 — Directional Derivatives**

**Definition**

- Directional derivative measures:

  - Rate of change of function along ANY chosen direction.

**Mathematical Form**

$$D_
uf=∇f⋅u$$

**where:**

- u = unit direction vector

**Geometric Interpretation**

- Directional derivative answers:

  - “How fast does function change if I move in THIS direction?”

**Projection Insight**

- Directional derivative:

  - projects gradient onto chosen direction.


**Key Insight**
- Maximum directional derivative occurs along gradient direction.


**This explains why:**

- gradient is steepest ascent direction.

**Example**

Suppose:

$$∇f=[3,4]$$

Magnitude:$$\sqrt{3^2+4^2}	​
=5$$

**Maximum change:**

- occurs along gradient direction.

#**PART 5 — Chain Rule (CRITICAL)**

**Why Chain Rule Matters**

- Deep learning systems contain:

  - Nested compositions of functions.

**Example:**

Input
-  → Layer 1
-  → Activation
-  → Layer 2
-  → Output
-  → Loss

**To train networks:**

- gradients must flow backward.

**This is achieved using:**


**Chain Rule**
- Single Variable Chain Rule

**Suppose:**

$$y=f(g(x))$$


Then: $$\frac{dy}{dg}=
\frac{dg}{dx}$$
	​



**Multivariable Chain Rule**

- For neural networks:

  - variables interact through multiple layers.


**Chain rule propagates derivatives through computation graph.**

- Computational Graph Intuition
- x → hidden layer → output → loss

**Gradient flows backward:**

- layer by layer


**ML Insight**
- Backpropagation = repeated chain rule.

  - This is the mathematical engine behind deep learning.

**Neural Network Perspective**

- Neural networks learn through:

   - forward pass
   - loss computation
backward pass using chain rule

**Without chain rule:**

- deep learning cannot exist.

**Python Example**

```python
x = 2

g = x**2
y = g + 1

dy_dg = 1
dg_dx = 2*x

dy_dx = dy_dg * dg_dx

print(dy_dx)
```

#**PART 6 — Jacobian**

**Definition**

- Jacobian generalizes derivatives for:

  - Vector-valued functions.

**Suppose:**

$$F(x)=[f_1(x),f_2(x),...,f_m(x)]$$


**Jacobian contains:**

- all first-order partial derivatives.


**Jacobian Matrix**

$$J_{ij}=\frac{∂f_i}{​∂x_j}$$
	​

	​
**Shape**
- $$(output dimension) × (input dimension)$$

**This is extremely important in ML.**

Example

- Suppose:

  - 3 outputs
  - 2 inputs

**Jacobian shape:**

$$3 × 2$$



**Geometric Meaning**

- Jacobian describes:

  - local transformation behavior
  - stretching
  - scaling
  - rotation

**ML Applications**

- Used in:

  - backpropagation
  - vector transformations
  - normalizing flows
  - diffusion models
  - sensitivity analysis


**Python Example**

```python
import numpy as np

J = np.array([
    [2, 1],
    [0, 3]
])

print(J)
```

#**PART 7 — Hessian**

**Definition**

- Hessian captures:

  - Second-order derivatives.

**It measures:**

- curvature
- bending
- local shape

**Hessian Matrix**

$$H_{ij} = \frac{∂^2f}{∂x_i∂x_j}$$
	​


**Interpretation**

- Hessian tells:

  - whether surface curves upward/downward
  - optimization geometry
  - local landscape structure


**Surface Interpretation**
| Hessian Behavior   | Meaning      |
| ------------------ | ------------ |
| Positive curvature | Bowl-shaped  |
| Negative curvature | Dome-shaped  |
| Mixed curvature    | Saddle point |



**Optimization Insight**

- Positive definite Hessian:

  - local minimum

**Negative definite Hessian:**

- local maximum

**Indefinite Hessian:**

- saddle point

**Saddle Point Problem**

- Deep learning landscapes contain many:

  - saddle points
  - flat plateaus

Optimization becomes difficult.


**Newton’s Method**

- Uses:
  - gradients
  - Hessians

for faster optimization.

**Update:**

$$θ:=θ−H^−1∇f$$
