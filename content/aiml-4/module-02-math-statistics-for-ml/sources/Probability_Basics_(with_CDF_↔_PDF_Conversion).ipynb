**Probability is the mathematical language of uncertainty.**

Modern AI systems do not make “certain” decisions. They make probabilistic predictions.

**Every ML model answers questions like:**
- “How likely is spam?”
- “What is the probability this transaction is fraud?”
- “Which next token is most probable in an LLM?”
- “What is the likelihood a user clicks this recommendation?”

#####**Real ML Examples**

**1. Large Language Models (LLMs)**

- When ChatGPT predicts the next word:

   - P(next token∣previous context)


The model outputs a probability distribution over vocabulary tokens.

**Example:**
- Token	Probability
- "world"	0.41
- "everyone"	0.22
- "people"	0.18

The highest probability token is sampled/generated.


**2. Fraud Detection**

- Banks compute:

  - P(Fraud∣Transaction Features)


**Example:**
- 0.02 → Safe transaction
- 0.95 → Highly suspicious


**3. Recommendation Systems**
- Netflix / Amazon estimate:

  - P(User Likes Item)

Recommendation ranking is fundamentally probability ranking.

####**Types of Random Variables**


**1. Discrete Random Variable**

- A variable that takes countable values.

**Examples:**
- Dice roll
- Number of customers
- Number of emails
- Number of defective products


**Characteristics**
- Finite OR countably infinite values
- Probabilities assigned to exact values

**Examples:**
- $$P(X=3)$$

is meaningful.



**2. Continuous Random Variable**

- A variable that can take infinitely many values in an interval.

**Examples:**

- Height
- Weight
- Temperature
- Time
- Voltage

**Characteristics**

- Infinite possible values
- Exact probability at a single point is zero


**For continuous variables:**

- $$P(X=x)=0$$


**Instead:**

- $$P(a≤X≤b)$$



matters.

###**CDF (Cumulative Distribution Function)**

**The CDF works for BOTH discrete and continuous variables.**


**Definition:**
- $$F(x)=P(X≤x)$$



**Meaning:**
- Total accumulated probability up to x.

**Intuition**
- CDF answers:

**“How much probability has accumulated so far?”**



**Example**

- Suppose:

$$X∼U(0,1)$$



**Then:**

- $$F(x)=x$$


**for:**

$$0≤x≤1$$



**Meaning:**
- At x = 0.2 → 20% probability accumulated
- At x = 0.7 → 70% probability accumulated

#####**CDF Properties**


**1. Increasing Function**
$$F(a)≤F(b)$$



if:

$$a<b$$


Because probability accumulates.



**2. Range**


 $$0≤F(x)≤1$$



**3. Right Continuous**

- CDF does not jump unexpectedly from the right.


**4. Limits**

$$
lim_{x→−∞}
	​
F(x)=0$$


$$
lim_{x→+∞}
	​
F(x)=0$$

#**Covariance**

**Covariance measures:**

- “Do two variables move together?”


**Formula**
- $$Cov(X,Y)=E[(X−E[X])(Y−E[Y])]$$


**Interpretation Table**
- Value	Meaning
- Positive	Move together
- Negative	Move oppositely
- Zero	No linear relation


**Example**

Suppose:

- More study hours → higher scores

- Then covariance tends positive.


**Negative Example**

Suppose:

- More speed → less travel time

- Covariance tends negative.




**Extremely Important Insight**
$$Cov(X,Y)=0$$

- DOES NOT imply independence.



####**Example**


**Suppose:**

$$Y=X^2$$

- with symmetric distribution around zero.

- Covariance can become zero.

- But variables are clearly dependent.
